```python
import json

with open("data.json", "r", encoding="utf-8") as archivo:
    datos = json.load(archivo)

ventas = datos["ventas"]

print(ventas)
```

```python
ingresos_productos = {}

for venta in ventas:
    producto = venta["producto"]
    importe = venta["cantidad"] * venta["precio_unitario"]

    if producto in ingresos_productos:
        ingresos_productos[producto] += importe
    else:
        ingresos_productos[producto] = importe

print(ingresos_productos)
```

```python
mayor_ingreso = max(ingresos_productos.values())

productos_mas_rentables = []

for producto, ingreso in ingresos_productos.items():
    if ingreso == mayor_ingreso:
        productos_mas_rentables.append(producto)

print("Producto(s) con mayores ingresos:", productos_mas_rentables)
print("Ingresos máximos:", mayor_ingreso, "€")
```

```python
ventas_ana = []

for venta in ventas:
    if venta["vendedor"] == "Ana":
        ventas_ana.append(venta)

print(ventas_ana)
```

```python
total_ana = 0

for venta in ventas_ana:
    importe = venta["cantidad"] * venta["precio_unitario"]
    total_ana += importe

print("Ventas realizadas por Ana:", len(ventas_ana))
print("Total facturado por Ana:", total_ana, "€")
```

```python
total_facturado = 0

for venta in ventas:
    importe = venta["cantidad"] * venta["precio_unitario"]
    total_facturado += importe

ticket_medio = total_facturado / len(ventas)

print("Total facturado:", total_facturado, "€")
print("Número de ventas:", len(ventas))
print("Ticket medio:", ticket_medio, "€")
```
